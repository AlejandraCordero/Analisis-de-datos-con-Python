Auditoría y Visualización de Ventas

```python
#Importar librerias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Cargar los datos
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
df = pd.read_csv(url)
print(df)
```

```python
# 3. GroupBy (dia, propina)

group_dia = df.groupby('day')['tip'].sum()
print(group_dia)
```

```python
#opcional
group_dia = df.groupby('day')['tip'].agg(['sum','mean','count'])
print(group_dia)
```

```python
# 4. Ingreso por dia
group_dia['sum'].plot(kind='bar')
plt.xlabel('Dia')
plt.ylabel('Ingreso')
plt.title('Ingreso por dia')
plt.show()
```

```python
#Histograma de montos totales (total bill)
df['total_bill'].plot(kind='hist')
plt.show()
```
