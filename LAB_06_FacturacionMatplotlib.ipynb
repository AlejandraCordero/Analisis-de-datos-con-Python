```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print ("Dataset cargado")
print (df)
```

Detección y Limpieza de Valores Faltantes (Missing Values)


```python
# Valores nulos por columna
print(df.isna())
print(df.isna().sum())
```

```python
# Ejemplos
# Completar los datos 'mean' - promedio
# df['total_bill'].mean()
df['total_bill'] = df['total_bill'].fillna(df['total_bill'].mean())
df['tip'] = df['tip'].fillna(df['tip'].mean())
```

Limpieza de Cadenas de Texto (Strings)

```python
print(df['time'].unique())
df['time']=df['time'].str.strip().str.lower()
print(df)
print(df['time'].unique())
```

Creación de Gráficos Profesionales con Matplotlib


```python
consumo_genero = df.groupby('sex')['total_bill'].mean()
print(consumo_genero)

#grafico
genero = ['hombre', 'mujer']
valores = consumo_genero

#elementos del grafico
plt.title ('Consumoe de clientes por genero')
plt.xlabel('Genero')
plt.ylabel('Consumo')
plt.bar(genero, valores)
plt.show()

```
