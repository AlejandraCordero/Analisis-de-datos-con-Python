Aplicar los conceptos teóricos de Matplotlib (scatter, hist, boxplot y gráficos de barras) sobre un dataset de viviendas de Kaggle.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/housing.csv"

df = pd.read_csv(url)
print(df.head())
```

```python
# Definir la lista de nombres estándar de las columnas del dataset Boston Housing
column_names = [
    'CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE',
    'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV'
]

# Volver a cargar el dataset indicando que no tiene encabezado original y asignar la lista
df = pd.read_csv(url, header=None, names=column_names)

# Mostrar la lista de columnas y las primeras filas
print("Lista de columnas:", list(df.columns))
display(df.head())
```

Histograma de Frecuencia (hist) - MEDV Valores medios de vivienda o precios

```python
from matplotlib.typing import LineStyleType
plt.hist(df['MEDV'], bins=10, color='forestgreen', edgecolor='black')
plt.title('Histograma de Frecuencia - MEDV')
plt.xlabel('MEDV Precio mediano')
plt.ylabel('Frecuencia (cantidad casas)')
plt.grid(axis='y', linestyle= '--', alpha=1)
plt.show()
```

Gráfico de Dispersión (scatter) -MEDV y RM Cantidad de hab por precio

```python
plt.scatter(df['RM'],df['MEDV'],color='dodgerblue',alpha=0.6, edgecolors='k')
plt.title('Gráfico de Dispersión - MEDV y RM')
plt.xlabel('RM Cantidad de habitaciones')
plt.ylabel('MEDV Precio mediano')
plt.grid(axis='y', linestyle= '--', alpha=1)
plt.show()
```
