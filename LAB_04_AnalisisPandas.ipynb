Analisis de datos con Pandas utilizando Python.

```python
import pandas as pd
```

```python
df_titanic = pd.read_csv('Titanic.csv')
print(df_titanic)
```

```python
df_titanic.head(5)
```

```python
edades = df_titanic['Age']
print(edades.head(10))
```

```python
promedio_edad = df_titanic['Age'].mean()
print(f"El promedio de edad es: {promedio_edad}")
```

```python
# 1: salvado
# 0: fallecido
conteo = df_titanic['Survived'].value_counts()
print(conteo)
```

```python
df_ordenado = df_titanic.sort_values(by='Fare', ascending=False)
print(df_ordenado.head(10))
```
